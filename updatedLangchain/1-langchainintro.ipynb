#### 1. Define Agent and View Diagram

```python
from langchain.agents import create_agent
from langchain_core.tools import tool

@tool
def get_weather(location: str) -> str:
    """Get the current weather for a given location."""
    if "kolkata" in location.lower():
        return "The weather in Kolkata is currently rainy."
    return "sunny"

# Create the agent with the weather tool
agent = create_agent(
    model="google_genai:gemini-3.1-flash-lite",
    tools=[get_weather],
    system_prompt="You are a helpful assistant."
)
# Display diagram
agent
```

#### 2. Invoke the Agent and Display Output

```python
# agent.invoke({"messages": [{"role" : "user", "content" : "What is the weather in Kolkata?"}]})

response = agent.invoke({
    "messages": [("user", "What is the weather in Kolkata?")]
})

response["messages"][-1].content[0]["text"]
```
