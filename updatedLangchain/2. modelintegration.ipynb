### Streaming and Batch

#### Streaming
Most model can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chance as they are produced. You can use a loop to process each chunk in real time :

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()
os.environ["NVIDIA_API_KEY"] = os.getenv("NVIDIA_API_KEY")

model = init_chat_model("meta/muse-glimmer-30b", model_provider="nvidia")
```

```python
model.invoke("write me a 200 words paragraph on AI")
```

```python
for chunk in model.stream("write me a 200 words paragraph on AI"):
    # print(chunk.text)
    print(chunk.text, end = " | ", flush= True)
```

#### Batch
Batching a collection of independent requests wave model can significantly improve performance and reduce costs as the processing can be done in parallel.

```python
responses = model.batch([
    "what is quantum comupting?",
    "How can I travel in space?",
    "How to start Trading?"
])

for response in responses:
    print(response)
```

```python
responses = model.batch(
    ["what is quantum comupting?",
    "How can I travel in space?",
    "Why my don't fly after using airplane mode?"],
    config={
        "max_concurrency": 5, #limit to 5 parallel cell
    }
)
for response in responses:
    print(response)
```
