### Model Integratoin

```python
import os
from dotenv import load_dotenv

load_dotenv()

from langchain.chat_models import init_chat_model #intiaalize any kind of chat models
```

#### API Keys


```python
os.environ["NVIDIA_API_KEY"] = os.getenv("NVIDIA_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
# os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

#### Nvidia Model

```python
model = init_chat_model("meta/muse-glimmer-30b", model_provider="nvidia")

response = model.invoke("Hello! How are you?")
response.content
```

#### Gemini Model

```python
models = init_chat_model("google_genai:gemini-3.1-flash-lite")

response = models.invoke("Why do parrot talk?")
response.content[0]["text"]
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model = "gemini-3.1-flash-lite")
response = model.invoke("Why do parrots talk?")
response.content[0]["text"]
```

#### OpenAI Model

```python
models = init_chat_model("gpt-4.1-mini")

response = models.invoke("Who are you?")
response.content  
```

```python
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model = "gpt-4.1-mini")
response = model.invoke("Hello who are you?")
response.content
```
